### OOPs Concept

```python
class Car:
    # Calling COnstructor method.
    # We need constructor to make an object of this class.
    def __init__(self,model,year,color, for_sale):
        self.model= model
        self.year= year
        self.color= color
        self.for_sale= for_sale
car1 = Car("BMW", 2020, "Black", True)

print(car1.model)
print(car1.year)
print(car1.color)
print(car1.for_sale)

```

```python
car2 = Car("Audi", 2021, "White", False)
print(car2.model)
```
