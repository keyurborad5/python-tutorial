### Functions

```python
def birthday_wish(name, age):
    print(f"Happy Birthday {name}! You are now {age} years old.")
# Example usage
birthday_wish("Alice", 30)  # Output: Happy Birthday Alice! You are
```

### Arguments
- Well there are multiple types of arguments that can be passed to the function.
  - Positional: as seen in above case, first argument should be name and then age.
  - Keyword: let say you donot remme what pos argu goes first. So you explocitly mention the keyword followed by '=' and its value. Eg. age=30
  - Default: there are some instance where some of the arguments remains sames all the time but there are special cases where we need to pass the custom argument value. Hence while defining the function mention the parameter=value so that if while calling the function, this perticular argument value is not passed then it should take this default value.

### Arbitary Arguments
- *args - allows you to pass multiple non-key args
    - Allows you pack all the arguments in a tuple, which can be unpacked inside the function to use as necessary
- **kwargs - allows you to pass multiple keyword-arguments

```python
def add(*args):
    print(type(args))  # This will show that args is a tuple
    total = 0
    for num in args:
        total += num
    return total
# Example usage
# This add function can take any number of arguments and return their sum
print(add(1, 2, 3,33,22,34,232))  # Output: 6
```

```python
# Exampple 2
def display(*args):
    print(type(args))  # This will show that args is a tuple

    for arg in args:
        print(arg, end=' ')
# Example usage
display("Hello", "World", 42, True)  # Output: Hello World
```

```python
# Exampple 3
def display(*args):
    print(type(args))  # This will show that args is a tuple

    for arg in args:
        if type(arg)== str:
            print(arg.upper(), end=' ')
            continue
        elif type(arg) == int:

            print("digit is: ",arg, end=' ')
            continue
        else:
            print(arg, end=' ')
# Example usage
display("Hello", "World", 42, 44,56, True,1212)  # Output: HELLO WORLD 42 True
```

```python
# **kwargs
def display_kwargs(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")
# Example usage
display_kwargs(name="Alice", age=30, city="New York")  # Output:
```

```python
# Example of using *args and **kwargs
def example_function(*args, **kwargs):
    print("Positional arguments:", args)
    print("Keyword arguments:", kwargs) 
    for arg in args:
        print(arg, end=' ')
    print()
    for key, value in kwargs.items():
        print(f"{key}: {value}")  
# Example usage
example_function(1, 2, 3, name="Alice", age=30)
```

### MAtch case

```python
def is_weekend(day):
    match day.lower():
        case "saturday" | "sunday":
            return True
        case _:
            return False
# Example usage
print(is_weekend("Sunday"))  # Output: True
```

### Exception handling

```python
try:
    num= float(input("Enter a number: "))
    # Code that may raise an exception
    result = 10 / num  # This will raise a ZeroDivisionError
    print(f"Result: {result}")
except ZeroDivisionError:
    print("Error: Division by zero is not allowed.")
# except ValueError:
#     print("Error: Invalid input. Please enter a valid number.")
except Exception as e:
    print(f"An unexpected error occurred: {e}")
finally:
    print("Execution completed.")
```

### Multithreading

```python
import time
import threading

def task_1(first, last, time_):
    time.sleep(time_)
    print(f"Task 1 completed in {time_} seconda by { first } { last  }")
def task_2():
    time.sleep(5)
    print("Task 2 completed.")
def task_3():
    time.sleep(3)
    print("Task 3 completed.")  

chor1 = threading.Thread(target=task_1, args=("John", "Doe", 4))
chor1.start()
chor2 = threading.Thread(target=task_2)
chor2.start()
chor3 = threading.Thread(target=task_3)
chor3.start()
chor4 = threading.Thread(target=task_1, args=("Jane", "Smith",1))
chor4.start()
print("All tasks started.")
chor1.join()
chor2.join()
chor3.join()
chor3.join()

print("all tasks completed.")
```
