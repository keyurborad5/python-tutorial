#### Typecasting
- to change the data type of the variable from one type to another

```python
# Data type of a variable
print(type(first_name))  # Output: <class 'str'>
print(type(age))         # Output: <class 'int'>
print(type(height))      # Output: <class 'float'>
```

```python
# #### Typecasting
# Typecasting examples
age_str = str(age)  # Convert integer to string
print(f'the age_str val is {age_str} and it data type is',type(age_str))  # Output: <class 'str'>
height_int = int(height)  # Convert float to integer
print(f'the height valye is {height_int} and its datatype is ',type(height_int))  # Output: <class 'int'>
age_float = float(age)  # Convert integer to float
print(f'the age float value is {age_float} and its data typoe os ',type(age_float))  # Output: <class 'float'>
```

```python
str= "10.4"  # String representation of a float
age_str = float(str)  # Convert string to integer
print(f'the age_str value is {age_str} and its data type is',type(age_str))  # Output: <class 'int'>
```

### Input function 

```python
name= input("Enter your name: ")  # This will prompt the user to enter their name
age = int(input("Enter your age: "))  # This will prompt the user to enter their age
age = float(age)  # Convert the input age to an integer
age=age + 1  # Increment the age by 1

print(f'Hello, {name}!')  # Output: Hello, <user's name>!
print(f'Next year, you will be {age} years old!')  # Output: Next year, you will be <user's age + 1> years old!
```

### Arithmatic and Math

```python
# Arethematic
a= 10
b= 20   
print(f'The sum of {a} and {b} is {a + b}')  # Output: The sum of 10 and 20 is 30
print(f'The difference between {b} and {a} is {b - a}')  # Output: The difference between 20 and 10 is 10
print(f'The product of {a} and {b} is {a * b}')  # Output: The product of 10 and 20 is 200
print(f'The quotient of {b} divided by {a} is {b / a}')  # Output: The quotient of 20 divided by 10 is 2.0
print(f'The remainder of {b} divided by {a} is {b % a}')  # Output: The remainder of 20 divided by 10 is 0
print(f'The exponent of {a} raised to the power of {b} is {a ** b}')  # Output: The exponent of 10 raised to the power of 20 is 100000000000000000000 
```

```python
# Round function
print("Round function examples:")
print(round(3.14159, 2))  # Output: 3.14
print(round(2.71828, 3))  # Output: 2.718

# Absolute function
print("Asolute function examples:")
print(abs(-5))  # Output: 5
print(abs(3.14))  # Output: 3.14

# Max and Min functions
print("Max and Min function examples:")
print(max(10, 20, 30))  # Output: 30
print(min(10, 20, 30))  # Output: 10
# Power function
print("Power function example:")    
print(pow(2, 3))  # Output: 8 (2 raised to the power of 3)
```

```python
# MATH module
import math
# Using the math module
print("Math module examples:")
print(math.sqrt(16))  # Output: 4.0 (Square root of 16)
print(math.pi)  # Output: 3.141592653589793 (Value of pi)
print(math.e)  # Output: 2.718281828459045 (Value of e) 
print(math.factorial(5))  # Output: 120 (Factorial of 5)
# Using the math module for trigonometric functions
print(math.sin(math.pi / 2))  # Output: 1.0 (Sine of 90 degrees)
print(math.cos(0))  # Output: 1.0 (Cosine of    0 degrees)
print(math.tan(math.pi / 4))  # Output: 1.0 (Tangent of 45 degrees) 
# Using the math module for logarithmic functions
print(math.log(100,10))  # Output: 2.0 (Logarithm base 10 of 100)
print(math.log2(8))  # Output: 3.0 (Logarithm base 2 of 8)      
# Using the math module for exponential functions
print(math.exp(1))  # Output: 2.718281828459045 (e raised to the power of 1)
print(math.pow(2, 3))  # Output: 8.0 (2 raised to the power of 3)
# Using the math module for rounding functions 
print(math.ceil(3.14))  # Output: 4 (Ceiling of 3.14)
print(math.floor(3.14))  # Output: 3 (Floor of 3.14)
```
