```python
## Sample code to validate a username
while True:  # This will run indefinitely until a break statement is encountered
    username = input("Enter your username: ")  # This will prompt the user to enter their username
    if len(username) > 12:  # Check if the username length is greater than 12
        print('Username length greater than 12. Please enter a new username.')
    elif ' ' in username:  # Check if the username contains a space
        print('Username contains space. Please enter a new username.')
    elif not username.isalpha():  # Check if the username contains special characters
        print('Username contains special characters. Please enter a new username.')
    else:
        print('Username is valid')
        break  # Exit the loop if the username is valid
```

```python
## String Indexing Example
num="1234-567-890"
# Accessing Elemets in a string using indexing 
#[start(incl) : end(exclu) : step]
print(num[0:4])
print(num[5:8])  # Output: 567 (Accessing characters from index 5 to 7)
print(num[1:6:2])
print(num[::2])  # Output:  (Accessing every second character)
print(num[::])  # Output: 1234-567-890 (Accessing all characters in the string)
print(num[---2])  # Output: 0 (Accessing the last character)
print(num[-3:])  # Output: 890 (Accessing the last three characters)
print(num[-4:-1])  # Output: 89 (Accessing characters from index
print(num[::-1])  # Output: 098-765-4321 (Reversing the string)
```

### For Loop
- Usually used to get iterated for fixed amount of time, over Range, String , Sequence etc.

```python
for i in range(5):  # This will iterate from 0 to 4
for i in range(1, 11):  # This will iterate from 1 to 10
    print(i)  # Output: 1, 2, 3, ..., 1
for i in range(1, 11, 2):  # This will iterate from 1 to 10 with a step of 2
for i in reversed(range(1, 11)):  # This will iterate from 10 to 1

# We can also iterate over a characters of the string
for char in "Hello":
    print(char)  # Output: H, e, l, l, o (Iterating over each character in the string)

```

```python
fruits = ["apple", "banana", "cherry"] # List of fruits
print(fruits[0])  # Output: apple (Accessing the first element of the list)
# Here as we can see, we can access the elements of the list using indexing
# fruits[start(incl) : end(exclu) : step]
print(fruits[0:2])  # Output: ['apple', 'banana'] (Accessing elements from index 0 to 1)
print(fruits[1:])  # Output: ['banana', 'cherry'] (
print(fruits[::-1])  # Output: ['apple', 'banana'] (Accessing elements from the start to index 1)
# iterate with for loop
for fruit in fruits:
    print(fruit)  # Output: apple, banana, cherry (Iterating over each element in the list)
```

```python
# Other functiosn
print("List functions examples:")
print(len(fruits))  # Output: 3 (Length of the list)    
print(fruits.count('apple'))  # Output: 1 (Count occurrences of 'apple' in the list)
print(fruits.index('banana'))  # Output: 1 (Index of 'banana' in the list)
fruits.append('orange')  # Add 'orange' to the end of the list 
print(fruits)  # Output: ['apple', 'banana', 'cherry', 'orange'] (List after appending 'orange')
fruits.insert(1, 'kiwi')  # Insert 'kiwi' at index 1
print(fruits)  # Output: ['apple', 'kiwi', 'banana', '      
fruits.sort()  # Sort the list in ascending order
print(fruits)  # Output: ['apple', 'banana', 'cherry', 'kiwi', 'orange'] (List after sorting)
fruits.reverse()  # Reverse the order of the list       
print(fruits)  # Output: ['orange', 'kiwi', 'cherry', 'banana', 'apple'] (List after reversing)   
```

### 2D List

```python
Matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]  # Example of a 2D list (matrix)
print("2D List (Matrix) examples:")
for row in Matrix:
    for element in row:
        print(element, end=' ')  # Output: 1 2 3 4 5 6 7 8 9 (Iterating over each element in the matrix)
    print()  # New line after each row
```

```python
# Indexing in 2D List
print("Accessing elements in 2D List (Matrix):")
print(Matrix[0][0])  # Output: 1 (Accessing the first element of the first row)
print(Matrix[1][2])  # Output: 6 (Accessing the third element
print(Matrix[2][:])  # Output: 8 (Accessing the second element of the third row)
print(Matrix[0:][1])  # Output: 8 (Accessing the second element of the third row)

```

### Dictionaries
- Collection of {key:value} pairs
- Ordered and Changable but not duplicate

```python
cap ={"USA": "United States of America", "UK": "United Kingdom", "IN": "India"}  # Example of a dictionary
print(dir(cap))  # Output: List of methods and attributes of the dictionary
```

```python
# Examples of each in dictionary 'clear', 'copy', 'fromkeys', 'get', 'items', 'keys', 'pop', 'popitem', 'setdefault', 'update', 'values'
# Indexing in Dictionary
print("Accessing elements in Dictionary:")
print(cap["USA"])  # Output: United States of America (Accessing the value associated with the key 'USA')
print(cap.get("UK"))  # Output: United Kingdom (Using get method to access value
print(cap.items())  # Output: dict_items([('USA', 'United States of America'), ('UK', 'United Kingdom'), ('IN', 'India')]) (Getting all key-value pairs)
print(cap.keys())  # Output: dict_keys(['USA', 'UK', 'IN'])
print(cap.values())  # Output: dict_values(['United States of America', 'United Kingdom', 'India']) (Getting all values)
print(cap.pop("IN"))  # Output: India (Removing the key 'IN' and returning its value)
print(cap)  # Output: {'USA': 'United States of America', 'UK': 'United Kingdom'} (Dictionary after popping 'IN')
print(cap.popitem())  # Output: ('UK', 'United Kingdom') (Removing and returning the last key-value pair)
print(cap)  # Output: {'USA': 'United States of America'} (Dictionary after popping the last item)
print(cap.setdefault("CA", "Canada"))  # Output: Canada (Setting default value for key 'CA' if it doesn't exist)
print(cap)  # Output: {'USA': 'United States of America', 'CA': 'Canada'} (Dictionary after setting default value for 'CA')
cap.update({"AU": "Australia"})  # Updating the dictionary with a new key-value pair
print(cap)  # Output: {'USA': 'United States of America', 'CA': 'Canada', 'AU': 'Australia'} (Dictionary after updating with 'AU')
```

```python
# For loop in Dictionary
print("Iterating over Dictionary:")
for key, value in cap.items():
    print(f'{key}: {value}')  # Output: USA: United States of America, UK: United Kingdom, IN: India (Iterating over each key-value pair)
```
