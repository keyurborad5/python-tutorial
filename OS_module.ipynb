#### Linux and Windows Paths
On Windows, paths are written using backslashes ('\\') as the separator between folder names. On Unix based operating system such as macOS, Linux, and BSDs, the forward slash ('/') is used as the path separator. Joining paths can be a headache if your code needs to work on different platforms.

```python
import os
print(os.path.join("folder", "subfolder", "file.txt"))
path=".\\UMDCP\\Sem4\\ENPM809\\"
print(os.path.join(path,"Assignment1.docs"))
```

```python
for file in my_files:
    if file.endswith('.txt'):
        print(os.path.join('C:\\users\\Keyur', file))
    if file.endswith('.mp4'):
        print(os.path.join('D:\\users\\Borad', file))
```

### Path and File validity
- Checking if a file/directory exists

```python
import os

print(os.path.exists('.'))
print(os.path.exists('test_dir'))
```

- Checking if a path is a file

```python
print(os.path.isfile('car_class.py'))
print(os.path.isfile('car.py'))
```

- Checking if a path is a directory

```python
print(os.path.isdir('test_dir'))
print(os.path.isdir('car_class.py'))
```

- Listing directories
     - This is the way we can list all the files inside a directory and then we can iterate through each file and check its file extension using for loop and .endswith() method

```python
print(os.listdir('.'))
my_files= os.listdir('.')
for file in my_files:
    if file.endswith('.ipynb'):
        print(file) # printts only python notebook files
```

- Stats of the directory/ file

```python
from datetime import datetime
print(os.stat('renamed_test_dir'))
print(os.stat('renamed_test_dir').st_size)  # Size in bytes
print(os.stat('renamed_test_dir').st_mtime)  # Last modified time

# In human readable format
print(datetime.fromtimestamp(os.stat('renamed_test_dir').st_mtime))
```

### Walk through each files and subfolders from a directory 

```python
for direc, folders, files in os.walk('.'):
    print(f"Directory: {direc}")
    for folder in folders:
        print(f"  Subfolder: {folder}")
    for file in files:
        print(f"  File: {file}")
```
