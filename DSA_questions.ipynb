### HashMaps or Dictionary.
- Two sum 
- nums=[1,5,22,3,11] Target=[14]
- Out=[3,4]

```python
def two_sum(nums, target):
    """
    Find two numbers in the list that add up to the target value.

    :param nums: List of integers
    :param target: Target sum
    :return: Tuple of the two numbers that add up to the target, or None if no such pair exists
    """
    dict={}
    for num in nums:
        compliment = target - num
        if compliment in dict:
            return (dict.get(compliment), nums.index(num))
        dict[num] = nums.index(num)     
    return None

nums = [2, 7, 11, 15]
target = 17
result = two_sum(nums, target)
if result:
    print(f"Indices of the two numbers that add up to {target}: {result}")   
```
