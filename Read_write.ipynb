#### Read and Write Files

```python
with open('log2.log') as logs:
    print(type(logs.read())) # .read() reads the entire content of file as a one string
```

```python
with open('log2.log') as logs:
    print(type(logs.readlines())) # .readlines() reads the entire content of file as a list of strings
```

```python
with open('log2.log') as logs:
    for line in(logs.readlines()): # .readlines() reads the entire content of file as a list of strings
        if '[warn]' in line:
            print(line)
```

```python
with open('log2.log') as logs:
    print((logs.read(100)), end='-----') # .read() reads the entire content of file as a one string
    print((logs.read(100)), end='------') # .read() reads the entire content of file as a one string
    
    print((logs.read(100)), end='-----') # .read() reads the entire content of file as a one string
```

### Write file
- if the file doesn't exist, it will be created annd if exists it will overwrite the content of the file

```python
with open('test.txt', 'w') as wf: # if the file doesn't exist, it will be created annd if exists it will append to the content
    wf.write("Hello World!") # write hello world
```

```python
with open('test.txt', 'w') as wf:
    wf.write("Nice to meet you!")
    wf.write("did you notice ?, you hello world got erased")
    wf.write("\n") # unless you give backward slash n it wont start from new line
    wf.write("also it does not start with new line")
    print(dir(wf))
```

```python
with open('test.txt', 'w') as wf:
    wf.writelines("Nice to meet you!")
    wf.writelines("did you notice ?, you hello world got erased")
    wf.write("\n") # unless you give backward slash n it wont start from new line
    wf.writelines("Hows it going")
    wf.writelines("Lovely Day")
    wf.write("\n") # unless you give backward slash n it wont start from new line

    wf.writelines("lets go home")

    print(wf.name)
```

### Can make a exact copy of a file

```python
with open('test.txt', 'r') as rf:
    with open('cpoy_test.txt', 'w') as wf:
        for line in rf.readlines(): # 
            
            wf.write(line)
```

- Above method copies line wise, sometime you never know the size of each line or may be the whole document can be of 1 line
- Hence its better to copy character wise

```python
with open('test.txt', 'r') as rf:
    with open('cpoy_test.txt', 'w') as wf:
        conten= rf.read(10)
        while len(conten)>0:
            wf.write(conten)
            conten= rf.read(10)
```

- Similarly can also be done for the image
    - Just you need to change the mode type - for reading 'r' to 'rb'
    - and writing w to 'wb'

```python
with open('Kinstugi.jpeg', 'rb') as rf:
    with open('cpoy_kins.jpeg', 'wb') as wf:
        chunks= rf.read(10)
        while len(chunks)>0:
            wf.write(chunks)
            chunks= rf.read(10)
```
